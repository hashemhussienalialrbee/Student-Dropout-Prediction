# 5. Outliers Detection

```python
# Detect outliers using the IQR method.
# Outliers are investigated rather than automatically removed, because extreme values may represent valid observations.

numerical_cols = ['Age', 'Family_Income', 'Study_Hours_per_Day',
                  'Attendance_Rate', 'Assignment_Delay_Days', 'Travel_Time_Minutes',
                  'Stress_Index', 'GPA', 'Semester_GPA', 'CGPA']
outlier_results = []

for i in numerical_cols:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[i] < lower_bound) | (df[i] > upper_bound)]

    outlier_results.append({
        'Column': i,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'max': df[i].max(),
        'min': df[i].min(),
        'Number of Outliers': len(outliers),

    })

outlier_table = pd.DataFrame(outlier_results)

outlier_table
```

# 6. Cleaning Decision

6.1 Fill missing (Family_Income, Study_Hours_per_Day, Stress_Index, Parental_Education)

```python
# Decision: Handle missing numeric values with the median.
# Reason: median is less sensitive to extreme values than the mean.

df_clean=df.copy()
numeric_cols = ["Family_Income", "Study_Hours_per_Day", "Stress_Index"]

for col in numeric_cols:
    median_value = df_clean[col].median()
    print(f"{col}: filling missing values with median = {median_value:.2f}")
    df_clean[col] = df_clean[col].fillna(median_value)
```

```python
# Decision: Handle missing categorical values with the mode.
# Reason: the mode is the most frequent observed category and preserves the column type.

mode_value = df_clean["Parental_Education"].mode()[0]
print(f"the mode value is:",{mode_value})
df_clean["Parental_Education"]=df_clean["Parental_Education"].fillna(mode_value)
print(f"Filling missing values in 'Parental_Education' with mode = {mode_value}")
```

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

6.2 Converting Age data type

```python
# Convert Age from float to integer because age is represented in whole years.

df_clean['Age'] = df_clean['Age'].astype(int)
```

```python
df_clean.info()
```

6.3 Cap extreme Family_Income

```python
# Decision: Cap Family_Income values at the IQR upper bound (73,800).
# Reason: Values above this threshold were treated as extreme values and capped to reduce their influence on the analysis and model.

df_clean["Family_Income"]=df_clean["Family_Income"].clip(upper= 73800.000)
print(df_clean["Family_Income"].max())
```

# 7. Before & After Cleaning

```python
summary = pd.DataFrame({
    "Metric": ["Rows", "Columns", "Total missing values", "Duplicate rows",
               "Max Family_Income"],
    "Before": [
        df.shape[0], df.shape[1], int(df.isnull().sum().sum()), int(df.duplicated().sum()),
        round(df["Family_Income"].max(), 2)
    ],
    "After": [
        df_clean.shape[0], df_clean.shape[1], int(df_clean.isnull().sum().sum()), int(df_clean.duplicated().sum()),
        round(df_clean["Family_Income"].max(), 2)
    ]
})
summary
```

```python
summary["Change"] = summary["After"] - summary["Before"]

def color_change(val):
    if val < 0:
        return "color: #16a34a; font-weight: bold"
    elif val > 0:
        return "color: #dc2626; font-weight: bold"
    return "color: gray"

styled = (
    summary.style
    .set_table_styles([
        {"selector": "th", "props": [("background-color", "#1e3a8a"),
                                     ("color", "white"),
                                     ("font-size", "14px"),
                                     ("text-align", "center")]},
        {"selector": "td", "props": [("text-align", "center"),
                                     ("font-size", "13px")]}
    ])
    .set_properties(subset=["Before"], **{"background-color": "#fee2e2", "color": "black"})
    .set_properties(subset=["After"],  **{"background-color": "#dcfce7", "color": "black"})
    .map(color_change, subset=["Change"])
    .format({"Before": "{:,.2f}", "After": "{:,.2f}", "Change": "{:+,.2f}"})
    .hide(axis="index")
)
styled
```

# 8. Data Aggregation Using GroupBy & Pivot Table
